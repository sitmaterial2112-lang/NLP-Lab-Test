```python

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity


documents = [
    "Machine learning is a branch of artificial intelligence.",
    "Deep learning uses neural networks for learning.",
    "Data science uses statistics and machine learning.",
    "Computer networks connect computers and exchange data.",
    "Artificial intelligence helps in solving complex problems."
]


vectorizer = TfidfVectorizer()
tfidf = vectorizer.fit_transform(documents)


query = input("Enter search query: ")

q = vectorizer.transform([query])


scores = cosine_similarity(q, tfidf)[0]


ranked = scores.argsort()[::-1]

print("\nSearch Results:")
for i in ranked:
    if scores[i] > 0:
        print("Document:", documents[i])
        print("Similarity:", round(scores[i], 3))
```
