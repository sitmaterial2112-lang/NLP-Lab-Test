```python

import nltk
from nltk.tokenize import sent_tokenize
from nltk.corpus import stopwords
from collections import Counter
from transformers import pipeline

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')

text = input("Enter news article: ")
sentences = sent_tokenize(text)
words = nltk.word_tokenize(text.lower())

stop = set(stopwords.words('english'))
freq = Counter(w for w in words if w.isalpha() and w not in stop)

scores = {}
for s in sentences:
    score = 0
    for w in nltk.word_tokenize(s.lower()):
        if w in freq:
            score += freq[w]
    scores[s] = score

extractive = sorted(scores, key=scores.get, reverse=True)[:2]


summarizer = pipeline("summarization",
                      model="sshleifer/distilbart-cnn-12-6")

abstractive = summarizer(text, max_length=80,
                         min_length=20, do_sample=False)

print("\nExtractive Summary:")
print(" ".join(extractive))

print("\nAbstractive Summary:")
print(abstractive[0]['summary_text'])
```
