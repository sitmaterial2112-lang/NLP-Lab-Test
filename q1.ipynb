```python

import re

reviews = [
    ["Samsung mobile is good.", "Late delivery of order.", "Order ORD12345 is delayed.", "Contact abc@gmail.com.", "Samsung battery is excellent."],
    ["iPhone camera is good.", "Order ORD23456 cancelled.", "Poor customer support.", "Product is damaged.", "Contact xyz@gmail.com."],
    ["Samsung screen is poor.", "Order ORD34567 arrived.", "Fast delivery service.", "Good product quality.", "Refund is pending."],
    ["iPhone battery is weak.", "Order ORD45678 is missing.", "Good packaging.", "Contact user@gmail.com.", "Poor quality product."],
    ["Samsung performance is good.", "Order ORD56789 delivered.", "Late delivery.", "Refund received.", "iPhone display is excellent."]
]


for i in range(5):
    with open(f"file{i+1}.txt", "w") as f:
        f.write("\n".join(reviews[i]))


pattern = input("Enter search pattern: ")

for i in range(5):
    filename = f"file{i+1}.txt"
    with open(filename) as f:
        for sentence in f:
            if re.search(pattern, sentence, re.I):
                sentence = re.sub(pattern, r"[\g<0>]", sentence, count=1, flags=re.I)
                print(filename, ":", sentence.strip())
```
