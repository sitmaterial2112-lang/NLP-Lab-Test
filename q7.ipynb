```python

import re
from collections import Counter

text = input("Enter text: ").lower()
words = re.findall(r'\b\w+\b', text)

unigram = Counter(words)
bigram = Counter()
trigram = Counter()

for i in range(len(words)):
    unigram[words[i]] += 0

    if i < len(words)-1:
        bigram[(words[i], words[i+1])] += 1

    if i < len(words)-2:
        trigram[(words[i], words[i+1], words[i+2])] += 1

print("Unigram:", unigram)
print("Bigram:", bigram)
print("Trigram:", trigram)

context = input("Enter sequence: ").lower().split()


print("Unigram Prediction:", unigram.most_common(1)[0][0])


prediction = "No prediction"
for gram in bigram:
    if gram[0] == context[-1]:
        prediction = gram[1]
        break
print("Bigram Prediction:", prediction)


prediction = "No prediction"
if len(context) >= 2:
    for gram in trigram:
        if gram[0] == context[-2] and gram[1] == context[-1]:
            prediction = gram[2]
            break
print("Trigram Prediction:", prediction)
```
