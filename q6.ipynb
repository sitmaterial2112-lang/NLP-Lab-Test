```python
from bs4 import BeautifulSoup
from spellchecker import SpellChecker
import re

f = open("raw.txt", "r", encoding="utf-8")
text = f.read()
f.close()

print("Raw Text:", text)

text = BeautifulSoup(text, "html.parser").get_text()

text = re.sub(r'http\S+|www\.\S+', '', text)

sentences = text.split(".")
new_text = []

for s in sentences:
    s = s.strip()
    if s != "" and s not in new_text:
        new_text.append(s)

text = ". ".join(new_text)

spell = SpellChecker()
words = text.split()
new_words = []

for w in words:
    new_words.append(spell.correction(w) or w)

text = " ".join(new_words)

text = " ".join(text.split())

f = open("cleaned.txt", "w", encoding="utf-8")
f.write(text)
f.close()

print("\nCleaned Text:", text)
print("Cleaned content saved to cleaned.txt")
```
